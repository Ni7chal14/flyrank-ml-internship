## 2. The question: decision, action, cost of a wrong call

### Research Question

Which webpages should be prioritized for refresh based on observable search performance signals?

### Decision

Determine which pages deserve manual review before investing time in updating content.

### Unit of Analysis

One webpage (one content page).

### Output

A ranked list of webpages with a priority score, confidence score, and recommended action (Refresh, Monitor, Protect, Expand, or Review).

### Action

The SEO or content team reviews the highest-priority pages first and decides which pages should be refreshed or monitored.

### Cost of a Wrong Recommendation

**False Positive:** A page is refreshed even though it didn't require attention, wasting time and resources.

**False Negative:** A genuinely declining page is ignored, resulting in further loss of search visibility, impressions, and traffic.

### Why ML?

The dataset contains many interacting signals such as impressions, CTR, average position, content age, engagement, and freshness. While simple rules can capture obvious cases, machine learning can discover more complex relationships between these features and prioritize pages more effectively.
